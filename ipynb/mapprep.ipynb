This nb is for checking the CRS and convert the building geometries to EPSG: 4326. (cuz it;s required for displaying GeoJSON if we use Mapbox GL JS)

```python
import geopandas as gpd

gdf = gpd.read_file("/Users/pesg/Documents/PilotStudy2026/assets/combined.geojson")
print(gdf.crs)  # check CRS if it's using the thing we need
```

```python
print(gdf.total_bounds)  # check min and max X and Y coordinates
```

```python
gdf = gdf.set_crs(epsg = 2927, allow_override = True)

gdf = gdf.to_crs(epsg = 4326)  # change 2927 to 4326 to be usable in mapbox

print(gdf.total_bounds)  # [-122.65152334   47.09405206 -122.3550547    47.33999793] success!
```

```python
gdf.to_file(
    "/Users/pesg/Documents/PilotStudy2026/assets/combinedNEW4326.geojson",
    driver = "GeoJSON"
)  # create new geojson file
```
