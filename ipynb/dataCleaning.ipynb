This notebook is for combining the Pierced County tax parcel polygon and the Pierced County improvement builtas dataset. 

The final cleaned data file is labeled as combined.geojson, it is located in our 'assets' folder. 

Its colomns contains:
    'Parcel Number', 
    'HVAC Description',
    'Year Remodeled',
    'Site_Addre',
    'geometry',
    'Physical Age'

The combined.geojson is filtered for residences and units 5-10.


```python
import pandas as pd
import geopandas as gpd
```

```python
# read files 
parcels = gpd.read_file("assets/parcels.shp")
ib = pd.read_csv("assets/ib.csv", dtype={"Parcel Number": str})
```

```python
# convert common fields to str
parcels["TaxParcelN"] = (
    parcels["TaxParcelN"].astype(str).str.strip()
)

ib["Parcel Number"] = (
    ib["Parcel Number"].astype(str).str.strip()
)
```

```python
# join
combined = parcels.merge(
    ib,
    left_on="TaxParcelN",
    right_on="Parcel Number",
    how="left"
)
```

```python
# filter for residence and 5-10 units only
combined = combined[
    (combined["TaxParce_1"].str.strip().str.lower() == "residence") &
    (combined["Units"] >= 5) & 
    (combined["Units"] <= 10) 
]
```

```python
# clean col 
combined = combined[['Parcel Number',
    'HVAC Description',
    'Year Built',
    'Year Remodeled',
    'Site_Addre',
    'geometry',
    'Physical Age'
    ]]
```

```python
# downlaod geojson
combined.to_file(
    "assets/combined.geojson",
    driver="GeoJSON"
)
```
